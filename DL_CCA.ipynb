```python
import os, glob, zipfile, yaml
import numpy as np
from PIL import Image, ImageDraw
from google.colab import drive
drive.mount("/content/drive")
DRIVE = "/content/drive/MyDrive/CCA2"; os.makedirs(DRIVE, exist_ok=True)

with zipfile.ZipFile("/content/fabrid.zip") as z:
    ent = [i for i in z.infolist() if "yolov" in i.filename.lower()][0]
    inner = z.extract(ent, "/content/nested")
ROOT = "/content/yolo"
with zipfile.ZipFile(inner) as z2: z2.extractall(ROOT)

SIZE = 256
SPLITS = {"train": "train", "val": "valid", "test": "test"}
cfg = yaml.safe_load(open(f"{ROOT}/data.yaml"))
fix = {"Stain": "Stains"}
defect_names = [fix.get(n.replace("Defect", "").strip().title(), n.replace("Defect", "").strip().title()) for n in cfg["names"]]
class_names = ["background"] + defect_names
print("Classes:", dict(enumerate(class_names)))

def build(folder):
    X, Y, skipped, empty = [], [], 0, 0
    for p in sorted(glob.glob(f"{ROOT}/{folder}/images/*.jpg")):
        img = Image.open(p).convert("RGB").resize((SIZE, SIZE), Image.BILINEAR)
        mask = Image.new("L", (SIZE, SIZE), 0); dr = ImageDraw.Draw(mask)
        lp = p.replace("/images/", "/labels/").rsplit(".", 1)[0] + ".txt"
        n_poly = 0
        if os.path.exists(lp):
            for line in open(lp):
                v = line.split()
                if len(v) < 7: skipped += 1; continue
                xy = list(map(float, v[1:]))
                pts = [(xy[i] * SIZE, xy[i + 1] * SIZE) for i in range(0, len(xy) - 1, 2)]
                dr.polygon(pts, fill=int(v[0]) + 1); n_poly += 1
        empty += (n_poly == 0)
        X.append(np.asarray(img, dtype=np.uint8)); Y.append(np.asarray(mask, dtype=np.uint8))
    print(f"{folder}: {len(X)} images | no defect polygon: {empty} | non-polygon lines skipped: {skipped}")
    return np.stack(X), np.stack(Y)

out = {}
for key, folder in SPLITS.items(): out["X_" + key], out["Y_" + key] = build(folder)
out["class_names"] = np.array(class_names)
np.savez_compressed(f"{DRIVE}/fabric_256.npz", **out)
counts = np.bincount(out["Y_train"].ravel(), minlength=5)
for n, c in zip(class_names, counts): print(f"{n:15s} {100 * c / counts.sum():6.2f} % of training pixels")
print("saved ->", f"{DRIVE}/fabric_256.npz")
```
